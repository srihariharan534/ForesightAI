# Data Cleaning
**Description**: Handle missing values, duplicates, invalid records, and parse dates.

```python
import pandas as pd
import numpy as np

# Load data
# df = pd.read_csv('../datasets/raw/data.csv')
```

```python
def clean_data(df):
    # Drop duplicates
    df = df.drop_duplicates()
    
    # Impute missing values for numeric columns with median
    num_cols = df.select_dtypes(include=[np.number]).columns
    df[num_cols] = df[num_cols].fillna(df[num_cols].median())
    
    # Forward fill categorical data
    cat_cols = df.select_dtypes(include=['object']).columns
    df[cat_cols] = df[cat_cols].fillna(method='ffill')
    
    return df

# df_cleaned = clean_data(df)
# df_cleaned.to_csv('../datasets/cleaned/cleaned_data.csv', index=False)
```
